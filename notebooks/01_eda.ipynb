```python
events.groupby("event")["transactionid"].apply(
    lambda x: x.notna().sum()
)
```

```python
user_activity = (
    events.groupby("visitorid")
    .size()
    .sort_values(ascending=False)
)

user_activity.describe()
```

```python
user_activity.head(20)
```

```python
user_activity.value_counts().sort_index().head(20)
```

```python
# 2. Item activity
item_activity = events.groupby("itemid").size().sort_values(ascending=False)
print('Item Description \n',item_activity.describe())
```

```python
# 3. User event types
user_event_types = events.groupby("visitorid")["event"].nunique()
print('User Event Types \n',user_event_types.value_counts().sort_index())
```

```python
# 4. Number of buyers
buyers = events.loc[
    events["event"] == "transaction",
    "visitorid"
].nunique()

buyers
```

```python
# 5. Top purchased products
purchased_items = (
    events.loc[events["event"] == "transaction"]
    .groupby("itemid")
    .size()
    .sort_values(ascending=False)
)

purchased_items.head(20)
```
